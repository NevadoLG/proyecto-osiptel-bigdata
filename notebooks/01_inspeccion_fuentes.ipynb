```python
import pandas as pd
import os

base = "/home/nevado/BigData_UPAO/osiptel_staging"
archivos = {
    "2.1 Internet Fijo":
        f"{base}/internet_fijo_2_1/2.1 CONEXIONES DE INTERNET FIJO.xlsx",
    "2.2 Fibra Óptica":
        f"{base}/fibra_optica_2_2/2.2. CONEXIONES DE INTERNET FIJO DE FIBRA ÓPTICA.xlsx",
    "2.3 Internet Fijo Distrital":
        f"{base}/internet_fijo_distrital_2_3/2.3 CONEXIONES DE INTERNET FIJO POR DISTRITO.xlsx",
    "4.1 Telefonía Fija":
        f"{base}/telefonia_fija_4_1/4.1. LÍNEAS FIJAS DE ABONADO EN SERVICIO.xlsx"
}
```

```python
for nombre, ruta in archivos.items():
    xls = pd.ExcelFile(ruta)
    print("=" * 70)
    print(nombre)
    print("Archivo:", os.path.basename(ruta))
    print("Tamaño:", round(os.path.getsize(ruta) / 1024 / 1024, 2), "MB")
    print("Hojas:", xls.sheet_names)
```

```python
ruta_21 = archivos["2.1 Internet Fijo"]
vista_21 = pd.read_excel(
    ruta_21,
    sheet_name="Dataset",
    header=None,
    nrows=10
)
vista_21
```

```python
diccionario_21 = pd.read_excel(
    ruta_21,
    sheet_name="Diccionario",
    header=None
)
diccionario_21.head(15)
```

```python
df_21 = pd.read_excel(
    ruta_21,
    sheet_name="Dataset",
    header=3
)
df_21.head()
```

```python
df_21 = df_21.loc[:, ~df_21.columns.astype(str).str.startswith("Unnamed")]
df_21.head()
```

```python
print("Filas:", df_21.shape[0])
print("Columnas:", df_21.shape[1])
print("\nColumnas:")
for columna in df_21.columns:
    print("-", columna)
print("\nTipos de datos:")
print(df_21.dtypes)
```

```python
print("Fecha mínima:", df_21["Mes"].min())
print("Fecha máxima:", df_21["Mes"].max())
print("\nValores nulos:")
print(df_21.isnull().sum())
```

```python
print("Duplicados exactos:", df_21.duplicated().sum())
```

```python
duplicados_21 = df_21[df_21.duplicated(keep=False)]
print("Total de filas involucradas en duplicados:", len(duplicados_21))
duplicados_21.sort_values(
    ["Mes", "Tecnologia", "Segmento"]
).head(20)
```

```python
for columna in [
    "Empresa",
    "Tecnologia",
    "Segmento",
    "Departamento",
    "Rango Velocidad de Bajada (nominal)"
]:
    print(f"{columna}: {df_21[columna].nunique()} valores únicos")
```
